```python
import matplotlib.pyplot as plt  
import numpy as np  
  
def plot_option_profit_loss(price_range, option_type1, option_type2, buy_or_sell1, buy_or_sell2, strike_price1, strike_price2, option_price1, option_price2, output_path):  
    """  
    绘制期权组合及单个期权的盈亏图，并保存为PDF文件。  
  
    参数:  
    price_range (list or numpy array): 价格范围。  
    option_type1, option_type2 (str): 期权类型，'call' 表示看涨，'put' 表示看跌。  
    buy_or_sell1, buy_or_sell2 (str): 买入或卖出，'buy' 表示买入，'sell' 表示卖出。  
    strike_price1, strike_price2 (float): 行权价。  
    option_price1, option_price2 (float): 期权价。  
    output_path (str): 保存PDF文件的路径。  
    """  
      
    # 转换为numpy数组以便进行数学运算  
    price_range = np.array(price_range)  
      
    # 计算单个期权的盈亏  
    def calculate_profit_loss(price, option_type, buy_or_sell, strike_price, option_price):  
        if option_type == 'call':  
            profit_loss = np.maximum(price - strike_price, 0) - option_price  
        elif option_type == 'put':  
            profit_loss = np.maximum(strike_price - price, 0) - option_price  
        else:  
            raise ValueError("Invalid option type. Use 'call' for call options and 'put' for put options.")  
          
        if buy_or_sell == 'sell':  
            profit_loss = -profit_loss  
          
        return profit_loss  
      
    # 计算期权组合的盈亏（这里简单相加，可以根据需要调整组合方式）  
    profit_loss_option1 = calculate_profit_loss(price_range, option_type1, buy_or_sell1, strike_price1, option_price1)  
    profit_loss_option2 = calculate_profit_loss(price_range, option_type2, buy_or_sell2, strike_price2, option_price2)  
    profit_loss_combined = profit_loss_option1 + profit_loss_option2  
      
    # 绘制图表  
    plt.figure(figsize=(10, 6))  
      
    # 绘制第一个期权的盈亏情况  
    plt.plot(price_range, profit_loss_option1, label=f'{option_type1.capitalize()} {buy_or_sell1.capitalize()} Option')  
      
    # 绘制第二个期权的盈亏情况  
    plt.plot(price_range, profit_loss_option2, label=f'{option_type2.capitalize()} {buy_or_sell2.capitalize()} Option')  
      
    # 绘制期权组合的盈亏情况  
    plt.plot(price_range, profit_loss_combined, label='Combined Option Profit/Loss', linewidth=2)  
      
    # 添加标题和标签  
    plt.title('Option Profit/Loss')  
    plt.xlabel('Underlying Price')  
    plt.ylabel('Profit/Loss')  
    plt.legend()  
    plt.grid(True)  
      
    # 保存图表为PDF文件  
    plt.savefig(output_path, format='pdf')  
      
    # 可选：显示图表（如果您还想在屏幕上查看图表）  
    # plt.show()  
      
    # 关闭图表以释放资源  
    plt.close()  
  
# 示例使用  
price_range = np.arange(3150, 3301, 1)  
option_type1 = 'call'  
option_type2 = 'put'  
buy_or_sell1 = 'buy'  
buy_or_sell2 = 'sell'  
strike_price1 = 3200  
strike_price2 = 3200  
option_price1 = 286.4  
option_price2 = 273.6  # 假设的第二个期权价格  
output_path = '/Users/hahaha/Downloads/option_profit_loss.pdf'  # 指定保存PDF文件的路径  
  
plot_option_profit_loss(price_range, option_type1, option_type2, buy_or_sell1, buy_or_sell2, strike_price1, strike_price2, option_price1, option_price2, output_path)


```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def black_scholes_price(S, K, T, r, sigma, option_type='put'):
    """
    计算布莱克-舒尔斯期权价格
    S: 标的资产价格
    K: 执行价格
    T: 到期时间（年）
    r: 无风险利率（年化）
    sigma: 波动率（年化）
    option_type: 'call' 或 'put'
    """
    d1 = (np.log(S / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    
    if option_type == 'call':
        price = S * norm.cdf(d1) - K * np.exp(-r * T) * norm.cdf(d2)
    elif option_type == 'put':
        price = K * np.exp(-r * T) * norm.cdf(-d2) - S * norm.cdf(-d1)
    else:
        raise ValueError("option_type 必须是 'call' 或 'put'")
    
    return price

def delta(S, K, T, r, sigma, option_type='put'):
    """
    计算 Delta
    """
    price = black_scholes_price(S, K, T, r, sigma, option_type)
    d1 = (np.log(S / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * np.sqrt(T))
    
    if option_type == 'call':
        delta_val = norm.cdf(d1)
    elif option_type == 'put':
        delta_val = norm.cdf(d1) - 1
    else:
        raise ValueError("option_type 必须是 'call' 或 'put'")
    
    return delta_val

# 参数设置
S_range = np.linspace(80, 120, 100)  # 标的资产价格范围
K = 100  # 执行价格
T = 1.0  # 到期时间（1年）
r = 0.05  # 无风险利率
sigma = 0.2  # 波动率

# 计算 Delta
deltas = [delta(S, K, T, r, sigma, option_type='put') for S in S_range]

# 绘制图像
plt.figure(figsize=(10, 6))
plt.plot(S_range, deltas, label='Delta of Far Month At-the-Money Put Option')
plt.xlabel('Stock Price (S)')
plt.ylabel('Delta')
plt.title('Delta of Far Month At-the-Money Put Option')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def black_scholes_delta(S, K, T, r, sigma, option_type='put'):
    """
    计算布莱克-舒尔斯期权的 Delta（针对看跌或看涨期权）
    """
    d1 = (np.log(S / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * np.sqrt(T))
    
    if option_type == 'call':
        delta_val = norm.cdf(d1)
    elif option_type == 'put':
        delta_val = norm.cdf(d1) - 1
    else:
        raise ValueError("option_type 必须是 'call' 或 'put'")
    
    return delta_val

# 参数设置
S = 100  # 标的资产价格
K = 100  # 执行价格（平值期权）
r = 0.05  # 无风险利率（年化）
sigmas = [0.1, 0.2, 0.3, 0.4]  # 不同的波动率值（年化）
T_range_months = np.linspace(0, 12, 100)  # 到期时间范围（月）

# 将月转换为年（因为布莱克-舒尔斯公式中的时间是以年为单位的）
T_range_years = T_range_months / 12

# 初始化一个字典来存储不同波动率下的 Delta 值
deltas_by_sigma = {sigma: [] for sigma in sigmas}

# 计算不同波动率和到期时间下的 Delta 值
for sigma in sigmas:
    for T in T_range_years:
        delta_val = black_scholes_delta(S, K, T, r, sigma, option_type='put')
        deltas_by_sigma[sigma].append(delta_val)

# 绘制图像
plt.figure(figsize=(12, 8))
for sigma, deltas in deltas_by_sigma.items():
    plt.plot(T_range_months, deltas, label=f'Sigma = {sigma:.1f}')
plt.xlabel('Time to Maturity (months)')
plt.ylabel('Delta')
plt.title('Delta of At-the-Money Put Option vs. Time to Maturity for Different Volatilities (0-12 months)')
plt.legend()
plt.grid(True)
plt.show()
```
