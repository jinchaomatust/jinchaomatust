```python
import yfinance as yf

# Define the ticker symbols for the indices
tickers = ['^DJI', '^GSPC', '^IXIC']  # Dow Jones, S&P 500, NASDAQ Composite

# Define the start and end dates
start_date = '2023-12-01'
end_date = '2023-12-31'

# Download the data for all three indices
data = yf.download(tickers, start=start_date, end=end_date)

# Extract the 'Adj Close' prices (adjusted closing prices)
closing_prices = data['Adj Close']

# Combine the data into one DataFrame (optional: rename columns)
closing_prices.columns = ['DJIA', 'S&P500', 'NASDAQ']

# Save the combined data to one CSV file
closing_prices.to_csv('indices_closing_prices.csv')

print("Combined CSV file has been saved as 'indices_closing_prices.csv'!")
```
