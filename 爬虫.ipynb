```python
import akshare as ak
import pandas as pd

# 获取上证指数（000001）的每日历史数据
df_000001 = ak.index_zh_a_hist(symbol="000001", period="daily", start_date="19900101")
df_000001['IndexSymbol'] = '000001'  # 添加一个列来标识指数

# 获取深证成指（000006）的每日历史数据
df_000006 = ak.index_zh_a_hist(symbol="000006", period="daily", start_date="19910403")  # 注意：深证成指从1991年开始
df_000006['IndexSymbol'] = '000006'  # 添加一个列来标识指数

# 合并数据框
combined_df = pd.concat([df_000001, df_000006], ignore_index=True)

# 定义输出路径和文件名
output_path = '/Users/hahaha/Downloads/实证20242214/index_hist_combined.csv'

# 将数据框保存到指定路径的CSV文件中
combined_df.to_csv(output_path, index=False, encoding='utf_8_sig')
```
