# 面板数据

```python
import pandas as pd
from linearmodels.panel import PanelOLS
import statsmodels.api as sm

# 读取数据
file_path = '/Users/hahaha/Downloads/panel data.csv'
df = pd.read_csv(file_path)

# 确保数据包含个体和时间索引
# 假设 CSV 文件中有名为 'entity' 的列表示个体，'time' 的列表示时间，'y' 是因变量，'x1', 'x2', ... 是自变量
df.set_index(['entity', 'time'], inplace=True)

# 分离因变量和自变量
y = df['y']
X = df.drop(columns=['y'])  # 假设除了 'y' 之外的所有列都是自变量

# 为了固定效应模型，我们需要给自变量添加常数项
X = sm.add_constant(X)

# 创建并拟合固定效应模型
model = PanelOLS(y, X, entity_effects=True)
results = model.fit()

# 输出回归结果
print(results)
```
